```python
from math import comb

p = [1, 1/2]

for d in range(2, 26):
    p.append((p[d - 1] + p[d - 2]) / 2)

print("Small cases:")
for d in range(10):
    print(f"d = {d}, p_d = {p[d]}")

answer_recurrence = p[25]

print("\nRecurrence result:")
print(answer_recurrence)

answer_formula = 2/3 + (1/3) * (-1/2) ** 25

print("\nClosed-form result:")
print(answer_formula)

answer_counting = sum(
    comb(25 - j, j) / 2 ** (25 - j)
    for j in range(13)
)

print("\nCombinatorial result:")
print(answer_counting)
print("\nDifferences:")
print(abs(answer_recurrence - answer_formula))
print(abs(answer_recurrence - answer_counting))
```
